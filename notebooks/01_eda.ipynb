Everything here is computed by step **`p1c_eda`** (`src/eda.py`) on the **full** train and test data

them — no analysis logic lives here.

```python
import os, sys
from pathlib import Path
import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents]
            if (p / "code/business_entity_resolution/src").is_dir())
sys.path.insert(0, str(ROOT / "code/business_entity_resolution/src"))
import eda

pd.set_option("display.max_colwidth", 70)
pd.set_option("display.max_rows", 200)
pd.set_option("display.max_columns", 30)

REPORT_DIR = Path(os.environ.get("EDA_REPORT_DIR", ROOT / "reports" / "p1c_eda"))
T = eda.load_tables(REPORT_DIR)
print(f"{len(T)} tables from {REPORT_DIR}")
eda.show(T, "summary", max_rows=None)
```

## 8 · Singleton traps

Nearest S2/S3 record by char-3gram TF-IDF name cosine, searched within the S1's own country. For a
singleton every candidate is wrong, so a high cosine is a trap. Matched S1s are the control group.
`nn owned by another S1` means exclusivity could reject the trap.

The 15 shown are the most trap-like of the sampled singletons, not a random 15.

```python
eda.show(T, "traps_stats", "traps_top15", max_rows=None)
```
